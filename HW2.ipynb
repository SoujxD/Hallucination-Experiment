```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("truthfulqa/truthful_qa", "generation", split="validation")
df = ds.to_pandas()

sampled = df.groupby("category").sample(n=2, random_state=42)
sampled = sampled.head(50)

sampled = sampled[[
    "question",
    "category",
    "best_answer",
    "correct_answers",
    "incorrect_answers"
]]

sampled.to_csv("truthfulqa_sample_50.csv", index=False)
```
